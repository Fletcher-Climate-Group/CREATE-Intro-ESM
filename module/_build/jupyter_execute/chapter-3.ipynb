# Model Intercomparisons and Future Climate Scenarios​

Learning Outcomes:​

1. Explain the concepts of climate sensitivity, radiative forcing and climate feedbacks ​

2. Describe what the Coupled Model Intercomparison Project (CMIP) is and its purpose ​

3. Understand what future emission scenarios are and how they are developed​

4. Explain the differences between predictions, forecasts and
projections​
