# Quantifying Uncertainty in Climate Simulations​

Learning Outcomes:​

1. Explain the concepts of internal climate variability, model
uncertainty and scenario uncertainty and how they contribute to uncertainty in future climate projections​

2. Be able to select the appropriate type of model ensemble for quantifying uncertainty​

3. Apply statistical methods to quantify different sources of uncertainty​

4. Understand how emergent constraints can be used to reduce
uncertainties in future climate projections​
